```python
from pathlib import Path
import pandas as pd

class Subject(object):

    def __init__(self, subject_id, bids_folder='/data/ds-valuecapture'):
        self.subject_id = subject_id
        self.bids_folder = Path(bids_folder)

    def get_behavioral_data(self, session=1):
        session_folder = self.bids_folder / 'sourcedata' / 'behavior' / f'sub-{self.subject_id}' / f'ses-{session}'

        if not session_folder.exists():
            raise FileNotFoundError(f"No behavioral data found for subject {self.subject_id} session {session}")

        df = []

        for run in range(1, 10):  # arbitrary max number of runs
            # sub-01_ses-1_task-val_cap_run-1_events.tsv
            #         # /Users/gdehol/data/ds-valuecapture/sourcedata/behavior/sub-01/ses-1/run-1
            # run_file = session_folder / f'run-{run}' / f'sub-{self.subject_id}_ses-{session}_task-val_cap_run-{run}_events.tsv'
            run_file = session_folder / f'run-{run}' / f'sub-{self.subject_id}_ses-{session}_task-val_cap_run-{run}_events.tsv'  # in case run folders are not used
            if run_file.exists():
                d = pd.read_csv(run_file, sep='\t')
                d['subject'], d['session'], d['run'] = self.subject_id, session, run
                df.append(d.set_index(['subject', 'session', 'run']))
            else:
                break

        return pd.concat(df)
```
