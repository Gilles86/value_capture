```python
from value_capture.utils.data import get_all_behavioral_data
import seaborn as sns
```

```python
df = get_all_behavioral_data()

tmp = df.groupby(['subject', 'condition', 'run'])['correct'].mean().reset_index()
sns.catplot(data=tmp, x='condition', y='correct', kind='point', hue='subject')
```

```python
tmp = df.groupby(['subject', 'condition'])['rt'].mean().reset_index()
sns.catplot(data=tmp, x='condition', y='rt', kind='point', hue='subject')
```
